```python
# Same repo setup as the other notebooks.
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "REPLACE_ME_WITH_GITHUB_URL_AFTER_PUSHING"

if IN_COLAB:
    if not os.path.exists("Gaza-Damage-Project"):
        subprocess.run(["git", "clone", "-b", "paper-aligned-pipeline", REPO_URL, "Gaza-Damage-Project"], check=True)
    os.chdir("Gaza-Damage-Project")

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "app/requirements.txt"], check=True)
```

```python
# Colab has no public localhost -- tunnel the Streamlit port out via
# `localtunnel` (no account needed, unlike ngrok).
import subprocess, threading, time, urllib.request

!npm install -g localtunnel -q

def run_streamlit():
    subprocess.run(["streamlit", "run", "app/streamlit_app.py", "--server.headless", "true", "--server.port", "8501"])

threading.Thread(target=run_streamlit, daemon=True).start()
time.sleep(8)

# print the public password (your Colab runtime's outbound IP) that
# localtunnel will ask visitors to confirm with
print("Tunnel password (your IP):", urllib.request.urlopen("https://loca.lt/mytunnelpassword").read().decode())
```
